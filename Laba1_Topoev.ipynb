## 1. Введение

В работе исследуется зависимость уровня звукового давления от пяти параметров аэродинамического эксперимента. Строятся модели Linear Regression, Ridge и Lasso. - `attack_angle` — угол атаки, градусы;
- `chord_length` — длина хорды, м;
- `free_stream_velocity` — скорость свободного потока, м/с;

Этапы работы: загрузить и исследовать данные; обработать пропуски; разделить выборку на обучающую и тестовую; оценить корреляции и VIF; построить три модели до PCA; проанализировать остатки; выбрать число компонент по порогу накопленной дисперсии 95%; повторить обучение и сравнить метрики.

## 2. Описание датасета

Airfoil Self-Noise опубликован в UCI Machine Learning Repository и содержит результаты аэродинамических и акустических испытаний профилей крыла. В наборе пять числовых входных признаков и одна числовая целевая переменная. Категориальных столбцов нет, поэтому кодирование не требуется.

```python
!pip -q install ucimlrepo

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

from IPython.display import display
from ucimlrepo import fetch_ucirepo
from sklearn.model_selection import train_test_split, KFold, cross_val_score
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.decomposition import PCA
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_percentage_error
from statsmodels.stats.outliers_influence import variance_inflation_factor
import statsmodels.api as sm

airfoil = fetch_ucirepo(id=291)
X_raw = airfoil.data.features.copy()
y_raw = airfoil.data.targets.iloc[:, 0].copy()
data = pd.concat([X_raw, y_raw], axis=1)
data.columns = ['frequency', 'attack_angle', 'chord_length', 'free_stream_velocity',
                'suction_side_displacement_thickness', 'scaled_sound_pressure']

print('Размер датасета:', data.shape)
display(data.head())
```

```python
print('Типы и количество значений:')
data.info()
print('\nПропущенные значения по столбцам:')
display(data.isna().sum().to_frame('Количество пропусков'))
print('Количество полных дубликатов:', data.duplicated().sum())
print('Описательная статистика:')
display(data.describe().T.round(3))
```

Проверка показала, что пропусков и полных дубликатов в датасете нет (0 пропусков, 0 дубликатов), поэтому заполнение и удаление строк не требуется. Описательная статистика показывает разброс каждого столбца: например, frequency варьируется от 200 до 20 000 Гц, а целевая переменная scaled_sound_pressure — от 103.38 до 140.99 дБ. Целевая переменная строго положительна, поэтому MAPE корректно определена и не имеет проблем с делением на ноль.

```python
fig, axes = plt.subplots(2, 3, figsize=(15, 8))
axes = axes.ravel()
for i, col in enumerate(data.columns):
    axes[i].hist(data[col], bins=30, edgecolor='black')
    axes[i].set_title(f'Распределение: {col}')
    axes[i].set_xlabel(col)
    axes[i].set_ylabel('Количество наблюдений')
    axes[i].grid(alpha=.25)
plt.suptitle('Распределения признаков и целевой переменной')
plt.tight_layout()
plt.show()
```

Гистограммы показывают, что распределения признаков далеки от нормальных: frequency принимает дискретные значения с несколькими пиками (мультимодальное распределение — следствие фиксированных частот эксперимента), attack_angle и suction_side_displacement_thickness заметно скошены вправо, а free_stream_velocity имеет два выраженных режима. Целевая переменная scaled_sound_pressure распределена относительно симметрично в диапазоне ~103–141 дБ. Такая форма распределений ожидаема для аэродинамических данных и не является основанием удалять наблюдения. Далее качество прогноза оценивается на отложенной тестовой выборке.

```python
plt.figure(figsize=(12, 6))
sns.boxplot(data=data)
plt.title('Диаграммы размаха признаков и целевой переменной')
plt.xlabel('Переменная')
plt.ylabel('Значение в исходных единицах')
plt.xticks(rotation=25)
plt.grid(axis='y', alpha=.25)
plt.tight_layout()
plt.show()
```

На диаграммах размаха заметнее всего выделяются большие значения frequency: диапазон частот доходит до 20 000 Гц, поэтому распределение этого признака существенно отличается от распределений остальных переменных. У suction_side_displacement_thickness также видны отдельные значения выше основной части наблюдений. При этом признаки измеряются в разных единицах и имеют разные масштабы, поэтому сравнивать высоту их диаграмм напрямую нельзя. Точки за пределами усов считаются потенциальными выбросами, но по одному графику нельзя определить, являются ли они ошибками измерения или отражают реальные условия эксперимента. Поэтому на данном этапе наблюдения не удаляются. Перед PCA признаки необходимо стандартизировать, чтобы переменные с большими числовыми значениями, например frequency, не доминировали только из-за масштаба.

```python
rows_before = len(data)
data = data.dropna().copy()
print('Удалено строк с пропусками:', rows_before - len(data))

features = ['frequency', 'attack_angle', 'chord_length', 'free_stream_velocity',
            'suction_side_displacement_thickness']
X = data[features]
y = data['scaled_sound_pressure']

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42
)

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print(f'Обучающая выборка: {X_train.shape[0]} строк')
print(f'Тестовая выборка: {X_test.shape[0]} строк')
print('Количество признаков:', X_train.shape[1])
```

Пропуски удаляются методом dropna; поскольку их нет, размер выборки не меняется (удалено 0 строк). Данные разделены в пропорции 80/20 с random_state=42 для воспроизводимости: 1202 строки в обучающей выборке и 301 строка в тестовой. StandardScaler настраивается только на обучающей части, после чего тем же преобразованием масштабируется тестовая часть — это исключает использование статистик тестовой выборки при обучении.

## 3. Корреляционный анализ и мультиколлинеарность

```python
corr = data[features + ['scaled_sound_pressure']].corr(method='pearson')
plt.figure(figsize=(10, 7))
sns.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm', center=0, vmin=-1, vmax=1)
plt.title('Матрица корреляций Пирсона')
plt.xlabel('Переменная')
plt.ylabel('Переменная')
plt.tight_layout()
plt.show()

corr_features = corr.loc[features, features].copy()
np.fill_diagonal(corr_features.values, 0)
strong_pair = corr_features.abs().stack().sort_values(ascending=False)
strong_pair = strong_pair[~strong_pair.index.map(lambda pair: pair[0] > pair[1])]
if len(strong_pair):
    pair = strong_pair.index[0]
    print(f'Наиболее сильная связь между признаками: {pair[0]} и {pair[1]}, r = {corr.loc[pair[0], pair[1]]:.3f}')
print('Корреляции признаков с целевой переменной:')
display(corr.loc[features, ['scaled_sound_pressure']].sort_values('scaled_sound_pressure', ascending=False).round(3))
```

Матрица корреляций показывает, что самая сильная линейная связь между входными признаками наблюдается у attack_angle и suction_side_displacement_thickness: r = 0.753. Это достаточно высокая положительная корреляция, поэтому при увеличении одного признака второй в этих данных также имеет тенденцию увеличиваться.

С целевой переменной scaled_sound_pressure сильнее всего связан frequency: r = −0.391. Связь отрицательная, то есть более высокие частоты в среднем связаны с более низким уровнем звукового давления. Для suction_side_displacement_thickness корреляция составляет −0.313, для chord_length — −0.236. У attack_angle связь слабее (−0.156), а у free_stream_velocity наблюдается небольшая положительная корреляция (0.125). Таким образом, ни один отдельный признак не демонстрирует сильной линейной связи с целевой переменной. Это помогает понять, почему для прогнозирования стоит проверить модель, использующую все пять признаков совместно. Корреляция при этом не означает причинно-следственную зависимость.

```python
X_vif = sm.add_constant(pd.DataFrame(X_train, columns=features), has_constant='add')
vif = pd.DataFrame({
    'Признак': features,
    'VIF': [variance_inflation_factor(X_vif.values, i + 1) for i in range(len(features))]
}).sort_values('VIF', ascending=False)
display(vif.round(3))
print('Максимальный VIF:', round(vif['VIF'].max(), 3))
```

Наибольший VIF получен для attack_angle — 3.509, затем идёт suction_side_displacement_thickness — 2.589. У остальных признаков значения ещё ниже: chord_length — 1.509, frequency — 1.144, free_stream_velocity — 1.037. Все показатели находятся ниже принятого порога 5. Несмотря на корреляцию attack_angle и suction_side_displacement_thickness (r = 0.753), расчёт VIF не показывает критичной мультиколлинеарности по выбранному критерию. Поэтому исходные признаки можно использовать для построения регрессионных моделей без обязательного удаления одного из них. PCA далее будет проверен как отдельный способ сокращения размерности.

## 4. Регрессионные модели до PCA

Сравниваются три модели: **Linear Regression** минимизирует сумму квадратов ошибок; **Ridge** добавляет L2-штраф к коэффициентам; **Lasso** добавляет L1-штраф и может уменьшать часть коэффициентов до нуля. Для Ridge и Lasso использованы `alpha=1.0` и `alpha=0.01` соответственно. Метрики на тестовой выборке рассчитываются один раз, а средний R² кросс-валидации — на пяти фолдах обучающей части. В кросс-валидации стандартизация помещена внутрь Pipeline, чтобы на каждом фолде она обучалась только на его тренировочной части.

```python
models = {
    'Linear Regression': LinearRegression(),
    'Ridge': Ridge(alpha=1.0),
    'Lasso': Lasso(alpha=0.01, max_iter=10000)
}
cv_splitter = KFold(n_splits=5, shuffle=True, random_state=42)
results_before = []
for name, model in models.items():
    model.fit(X_train_scaled, y_train)
    pred = model.predict(X_test_scaled)
    cv_pipeline = make_pipeline(StandardScaler(), model.__class__(**model.get_params()))
    cv_scores = cross_val_score(cv_pipeline, X_train, y_train, cv=cv_splitter, scoring='r2')
    results_before.append({
        'Модель': name,
        'CV R² (среднее)': cv_scores.mean(),
        'RMSE': np.sqrt(mean_squared_error(y_test, pred)),
        'R²': r2_score(y_test, pred),
        'MAPE, %': mean_absolute_percentage_error(y_test, pred) * 100
    })
results_before_df = pd.DataFrame(results_before)
display(results_before_df.round(4))
best_before_r2 = results_before_df.loc[results_before_df['R²'].idxmax()]
best_before_rmse = results_before_df.loc[results_before_df['RMSE'].idxmin()]
best_before_mape = results_before_df.loc[results_before_df['MAPE, %'].idxmin()]
print(f"Максимальный тестовый R²: {best_before_r2['Модель']} ({best_before_r2['R²']:.4f})")
print(f"Минимальный RMSE: {best_before_rmse['Модель']} ({best_before_rmse['RMSE']:.4f})")
print(f"Минимальный MAPE: {best_before_mape['Модель']} ({best_before_mape['MAPE, %']:.2f}%)")
```

Ridge и Lasso показали практически такие же результаты, как обычная линейная регрессия: разница в R² составляет менее 0.001. В этом эксперименте регуляризация не дала заметного улучшения качества на тестовой выборке. Это согласуется с тем, что расчёт VIF не выявил выраженной мультиколлинеарности, хотя само по себе не доказывает, что именно это стало причиной сходства результатов.

```python
fig, ax = plt.subplots(1, 3, figsize=(15, 4))
ax[0].bar(results_before_df['Модель'], results_before_df['RMSE'])
ax[0].set_title('RMSE до PCA'); ax[0].set_ylabel('RMSE, дБ')
ax[1].bar(results_before_df['Модель'], results_before_df['R²'])
ax[1].set_title('R² до PCA'); ax[1].set_ylabel('Коэффициент R²')
ax[2].bar(results_before_df['Модель'], results_before_df['MAPE, %'])
ax[2].set_title('MAPE до PCA'); ax[2].set_ylabel('MAPE, %')
for a in ax: a.set_xlabel('Модель'); a.tick_params(axis='x', rotation=20); a.grid(axis='y', alpha=.25)
plt.tight_layout(); plt.show()
```

На трёх графиках Linear Regression показывает лучший результат: минимальные RMSE (4.7041 дБ) и MAPE (2.96%) и максимальный R² (0.5583). Ridge и Lasso расположены почти на том же уровне, поэтому визуально различия между моделями небольшие. Значение R² = 0.5583 означает, что модель объясняет около 55.83% вариации уровня звукового давления на тестовой выборке относительно прогноза средним значением. При этом RMSE около 4.7 дБ показывает, что ошибки прогноза всё ещё заметны. Поэтому линейная регрессия является лучшей из трёх проверенных моделей, но её качество нельзя считать идеальным.

## 5. Анализ остатков до PCA

Остаток — разность фактического и предсказанного значения. По графику остатков относительно предсказаний проверяется, располагаются ли ошибки случайно вокруг нуля или образуют структуру (тренд, дугу, воронку). Этот анализ проводится после базовых моделей и до PCA, как требуется в постановке задачи.

```python
best_name = results_before_df.loc[results_before_df['R²'].idxmax(), 'Модель']
residual_model = models[best_name]
pred_resid = residual_model.predict(X_test_scaled)
residuals = y_test.to_numpy() - pred_resid

plt.figure(figsize=(9, 5))
plt.scatter(pred_resid, residuals, alpha=.6)
plt.axhline(0, linestyle='--', color='black')
plt.title(f'Остатки модели {best_name} до PCA')
plt.xlabel('Предсказанный уровень звукового давления, дБ')
plt.ylabel('Остаток: факт − прогноз, дБ')
plt.grid(alpha=.25); plt.tight_layout(); plt.show()
print('Модель для анализа остатков:', best_name)
print('Среднее остатков:', round(float(np.mean(residuals)), 4))
print('Стандартное отклонение остатков:', round(float(np.std(residuals, ddof=1)), 4))
```

На графике остатки расположены по обе стороны от нулевой линии, поэтому модель как завышает, так и занижает прогнозируемый уровень звукового давления. Среднее значение остатков составляет 0.0058 дБ — практически ноль, а стандартное отклонение равно 4.7119 дБ. При прогнозах примерно от 127 до 132 дБ заметны как положительные остатки до нескольких дБ, так и большие отрицательные ошибки; отдельные точки опускаются примерно до −15 дБ. Поэтому модель не одинаково точно предсказывает уровень звукового давления для всех наблюдений.

Выраженной дугообразной формы или чёткого тренда на графике не видно, следовательно, явных визуальных признаков нелинейной структуры нет. При этом разброс точек по вертикали неодинаков на разных участках диапазона прогнозов. Это может указывать на неоднородность дисперсии ошибок, но по одному графику нельзя однозначно утверждать, что гетероскедастичность присутствует. В целом график показывает, что средняя ошибка близка к нулю, однако отдельные прогнозы имеют существенные отклонения.

## 6. Снижение размерности методом PCA

PCA применяется к стандартизированным признакам обучающей выборки. Число компонент выбирается по правилу: взять минимальное количество компонент, при котором накопленная объяснённая дисперсия не меньше 95%. PCA обучается только на train, затем обученное преобразование применяется к test.

```python
pca_full = PCA()
pca_full.fit(X_train_scaled)
explained = pca_full.explained_variance_ratio_
cumulative = np.cumsum(explained)
n_components = int(np.argmax(cumulative >= 0.95) + 1)
print('Количество исходных признаков:', X_train.shape[1])
print('Минимальное число компонент для порога 95%:', n_components)
print(f'Накопленная объяснённая дисперсия: {cumulative[n_components-1]:.4f} ({cumulative[n_components-1]*100:.2f}%)')
```

Число компонент выбирается по правилу сохранения не менее 95% накопленной дисперсии. Первые три компоненты объясняют примерно 83% дисперсии признаков, поэтому этого недостаточно. После добавления четвёртой компоненты накопленная доля достигает 96.58%, и условие выполняется. Таким образом, пять исходных признаков заменяются четырьмя главными компонентами при сохранении большей части вариации исходных данных. Пятая компонента объясняет оставшиеся примерно 3.42% дисперсии, но для достижения выбранного порога 95% она не требуется.

```python
components = np.arange(1, len(explained) + 1)
fig, ax = plt.subplots(figsize=(9, 5))
ax.bar(components, explained, alpha=.65, label='Доля дисперсии компоненты')
ax.plot(components, cumulative, marker='o', linewidth=2, label='Накопленная дисперсия')
ax.axhline(.95, linestyle='--', label='Порог 95%')
ax.axvline(n_components, linestyle='--', label=f'Выбрано компонент: {n_components}')
ax.set_title('График каменистой осыпи и накопленной дисперсии')
ax.set_xlabel('Номер главной компоненты')
ax.set_ylabel('Доля объяснённой дисперсии')
ax.set_xticks(components); ax.set_ylim(0, 1.05); ax.grid(axis='y', alpha=.25); ax.legend()
plt.tight_layout(); plt.show()
```

На графике первые три компоненты в сумме объясняют около 83% дисперсии, что ниже установленного порога 95%. После добавления четвёртой компоненты накопленная доля увеличивается до 96.58%, поэтому вертикальная линия на графике правильно отмечает минимальное число компонент, необходимое для выполнения условия. Пятая компонента увеличивает сохранённую дисперсию ещё примерно на 3.42%, но для выбранного критерия её использование не требуется. Следовательно, сокращение с пяти признаков до четырёх компонент соответствует принятому правилу.

```python
pca = PCA(n_components=n_components)
X_train_pca = pca.fit_transform(X_train_scaled)
X_test_pca = pca.transform(X_test_scaled)
print('Размер X_train после PCA:', X_train_pca.shape)
print('Размер X_test после PCA:', X_test_pca.shape)
print(f'Доля дисперсии, сохранённая выбранными компонентами: {pca.explained_variance_ratio_.sum():.4f}')
```

После PCA обучающая выборка имеет размер 1202 × 4, а тестовая — 301 × 4. Число строк не изменилось: преобразование не удаляет наблюдения, а заменяет пять исходных признаков четырьмя главными компонентами. Выбранные компоненты сохраняют 96.58% дисперсии исходных признаков. Поскольку главные компоненты ортогональны, они линейно некоррелированы друг с другом. На следующем этапе можно проверить, насколько такое сокращение размерности повлияло на качество прогноза целевой переменной.

## 7. Модели после PCA и сравнение результатов

На компонентах повторно обучаются те же три алгоритма с теми же параметрами. Это позволяет сравнить влияние именно преобразования PCA, не меняя набор моделей и параметры регуляризации.

```python
models_pca = {
    'Linear Regression': LinearRegression(),
    'Ridge': Ridge(alpha=1.0),
    'Lasso': Lasso(alpha=0.01, max_iter=10000)
}
results_after = []
for name, model in models_pca.items():
    model.fit(X_train_pca, y_train)
    pred = model.predict(X_test_pca)
    # В CV стандартизация и PCA заново обучаются внутри каждого фолда
    cv_pipeline = make_pipeline(StandardScaler(), PCA(n_components=n_components), model.__class__(**model.get_params()))
    cv_scores = cross_val_score(cv_pipeline, X_train, y_train, cv=cv_splitter, scoring='r2')
    results_after.append({
        'Модель': name,
        'CV R² (среднее)': cv_scores.mean(),
        'RMSE': np.sqrt(mean_squared_error(y_test, pred)),
        'R²': r2_score(y_test, pred),
        'MAPE, %': mean_absolute_percentage_error(y_test, pred) * 100
    })
results_after_df = pd.DataFrame(results_after)
display(results_after_df.round(4))
best_after_r2 = results_after_df.loc[results_after_df['R²'].idxmax()]
best_after_rmse = results_after_df.loc[results_after_df['RMSE'].idxmin()]
best_after_mape = results_after_df.loc[results_after_df['MAPE, %'].idxmin()]
print(f"Максимальный тестовый R² после PCA: {best_after_r2['Модель']} ({best_after_r2['R²']:.4f})")
print(f"Минимальный RMSE после PCA: {best_after_rmse['Модель']} ({best_after_rmse['RMSE']:.4f})")
print(f"Минимальный MAPE после PCA: {best_after_mape['Модель']} ({best_after_mape['MAPE, %']:.2f}%)")
```

После PCA лучшей из трёх моделей остаётся Linear Regression: R² = 0.5350, RMSE = 4.8266 дБ, MAPE = 3.05%. Ridge и Lasso снова дают очень близкие результаты. При сравнении с исходной моделью R² снизился с 0.5583 до 0.5350, RMSE вырос с 4.7041 до 4.8266 дБ, а MAPE — с 2.96% до 3.05%. Следовательно, после сокращения размерности качество прогноза немного ухудшилось. Хотя четыре компоненты сохраняют 96.58% дисперсии исходных признаков, это не гарантирует сохранения всей информации, полезной именно для предсказания звукового давления.

```python
comparison = results_before_df[['Модель', 'CV R² (среднее)', 'RMSE', 'R²', 'MAPE, %']].merge(
    results_after_df[['Модель', 'CV R² (среднее)', 'RMSE', 'R²', 'MAPE, %']],
    on='Модель', suffixes=(' до PCA', ' после PCA')
)
display(comparison.round(4))
comparison['Изменение RMSE'] = comparison['RMSE после PCA'] - comparison['RMSE до PCA']
comparison['Изменение R²'] = comparison['R² после PCA'] - comparison['R² до PCA']
comparison['Изменение MAPE, п.п.'] = comparison['MAPE, % после PCA'] - comparison['MAPE, % до PCA']
print('Изменения после PCA (отрицательное RMSE/MAPE и положительное R² означают улучшение):')
display(comparison[['Модель', 'Изменение RMSE', 'Изменение R²', 'Изменение MAPE, п.п.']].round(4))
```

Таблица показывает одинаковое направление изменений для всех трёх моделей: после PCA RMSE и MAPE увеличились, а R² снизился. У Linear Regression RMSE вырос с 4.7041 до 4.8266 дБ, R² снизился с 0.5583 до 0.5350, а MAPE увеличился с 2.9646% до 3.0457%. У Ridge RMSE вырос с 4.7050 до 4.8272 дБ, а у Lasso — с 4.7082 до 4.8280 дБ. Снижение R² составляет около 0.023 для каждой модели. Таким образом, на этой тестовой выборке PCA не улучшил прогноз ни для одного из трёх алгоритмов. Результат показывает, что сохранение 96.58% дисперсии признаков не гарантирует улучшения регрессии: PCA ориентируется на вариацию самих признаков, а не непосредственно на качество предсказания целевой переменной.

```python
x = np.arange(len(comparison)); width = .35
fig, ax = plt.subplots(1, 3, figsize=(16, 4.5))
metric_specs = [
    ('RMSE до PCA', 'RMSE после PCA', 'RMSE до/после PCA', 'RMSE, дБ'),
    ('R² до PCA', 'R² после PCA', 'R² до/после PCA', 'Коэффициент R²'),
    ('MAPE, % до PCA', 'MAPE, % после PCA', 'MAPE до/после PCA', 'MAPE, %')
]
for a, (before_col, after_col, title, ylabel) in zip(ax, metric_specs):
    a.bar(x - width/2, comparison[before_col], width, label='До PCA')
    a.bar(x + width/2, comparison[after_col], width, label='После PCA')
    a.set_title(title); a.set_ylabel(ylabel); a.set_xlabel('Модель')
    a.set_xticks(x); a.set_xticklabels(comparison['Модель'], rotation=20)
    a.legend(); a.grid(axis='y', alpha=.25)
plt.tight_layout(); plt.show()
```

На графиках для каждой модели столбец RMSE после PCA немного выше, чем до преобразования, а столбцы R² и MAPE показывают ухудшение в соответствующем направлении: R² становится ниже, MAPE — выше. При этом разница между Linear Regression, Ridge и Lasso остаётся небольшой как до, так и после PCA. Наиболее заметный пример — Linear Regression: RMSE увеличился на 0.1225 дБ, а R² уменьшился на 0.0233. Значит, графики подтверждают таблицу: в данном эксперименте переход к четырём главным компонентам не дал улучшения качества предсказания.
